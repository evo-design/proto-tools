# ESMFold2 Inversion Gradient

Differentiate ESMFold2 inversion structure objectives with respect to a relaxed binder.


```python
from proto_tools.tools import (
    ESMFold2InversionCandidate,
    ESMFold2InversionGradientConfig,
    ESMFold2InversionGradientInput,
    run_esmfold2_inversion_gradient,
)
from proto_tools.utils import one_hot_protein_logits
```

```python
target = "MKTAYIAKQRQISFVKSHFSRQLEERLGLIEVQ"
binder = "MKTAYIAKQRQISFVKSHFSRQLE"

result = run_esmfold2_inversion_gradient(
    ESMFold2InversionGradientInput(
        inputs=ESMFold2InversionCandidate(
            logits=one_hot_protein_logits(binder, sharpness=2.0),
            target_sequence=target,
        )
    ),
    ESMFold2InversionGradientConfig(
        gradient_terms=["intra_contact", "inter_contact", "globularity"]
    ),
)

for term in result.results[0].terms:
    print(term.term.id, term.term_value)
```

Each quantity carries its own gradient. Combining them is the caller's choice:


```python
objective = result.results[0].compose(
    weights={result.results[0].terms[0].term: 0.5}
)
print(objective.objective_loss)
```
