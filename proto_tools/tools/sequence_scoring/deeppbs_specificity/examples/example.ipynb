# DeepPBS Specificity

![DeepPBS Specificity](https://proto-bio.github.io/proto-assets/images/tool/deeppbs_specificity/hero.png)

DeepPBS predicts the DNA base preferences of a protein directly from a protein-DNA complex structure. Given one or more PDB files, the `deeppbs-specificity` tool returns a canonical DNA-only position probability matrix (PPM) in `A,C,G,T` order, alongside the true DNA sequence, residue masks, and per-base chain labels.

This tool requires a local DeepPBS repository and a local X3DNA install. By default, a missing dependency or failed run raises; pass `allow_fallback=True` to instead return a uniform fallback PPM flagged with `used_fallback`.

## Available tools

```python
from proto_tools.utils.notebook_docs import display_available_tools

display_available_tools("deeppbs_specificity")
```

## Usage

Provide one or more protein-DNA PDB structures. Each result carries a `predicted_ppm` (`L x 4`, ACGT order), the `used_fallback` / `fallback_reason` markers, and an `output_npz_path` that is `None` unless `output_directory` is set.

```python
from pathlib import Path

from proto_tools import (
    DeepPBSSpecificityConfig,
    DeepPBSSpecificityInput,
    run_deeppbs_specificity,
)

# The Zif268 zinc finger-DNA co-crystal (1AAY) bundled with the toolkit.
pdb_path = str(Path("..") / "example_input_fixture.pdb")

inputs = DeepPBSSpecificityInput(pdb_paths=[pdb_path])
# allow_fallback=True returns a uniform fallback PPM (flagged via used_fallback)
# when DeepPBS deps are missing; omit it to raise instead.
config = DeepPBSSpecificityConfig(allow_fallback=True)

output = run_deeppbs_specificity(inputs, config)

result = output.results[0]
print("used_fallback:", result.used_fallback)
print("fallback_reason:", result.fallback_reason)
print("PPM shape:", len(result.predicted_ppm), "x", len(result.predicted_ppm[0]))
print("output_npz_path:", result.output_npz_path)
```
